```python
import os
import shutil


print("ani-cli:", shutil.which("ani-cli", path=os.path.expanduser("~/.local/bin")) or "não encontrado")

package_managers = {
    "pacman": "sudo pacman -Syu && sudo pacman -S --needed mpv yt-dlp",
    "apt": "sudo apt update && sudo apt install mpv yt-dlp",
    "dnf": "sudo dnf install mpv yt-dlp",
    "zypper": "sudo zypper install mpv yt-dlp",
}

available_manager = next(
    (manager for manager in package_managers if shutil.which(manager)),
    None,
)

if shutil.which("mpv") and shutil.which("yt-dlp"):
    print("mpv e yt-dlp já estão instalados.")
elif available_manager:
    print("mpv/yt-dlp não encontrados.")
    print("Execute no terminal normal (fora do notebook):")
    print(package_managers[available_manager])
else:
    print("Nenhum gerenciador de pacotes foi encontrado no ambiente do notebook.")
    print("Execute no terminal do sistema, por exemplo no Arch Linux:")
    print("sudo pacman -Syu && sudo pacman -S --needed mpv yt-dlp")
```

```python
import os
import platform
import shutil
import subprocess
from pathlib import Path


def command_output(command):
    executable = shutil.which(command[0])
    if not executable:
        return f"{command[0]}: não encontrado"
    result = subprocess.run(command, capture_output=True, text=True)
    output = (result.stdout + result.stderr).strip()
    return output or f"saída vazia (código {result.returncode})"


print("=== SISTEMA ===")
print(platform.platform())
print(command_output(["uname", "-a"]))
print("XDG_SESSION_TYPE:", os.environ.get("XDG_SESSION_TYPE", "não definido"))

print("\n=== REPRODUTORES E CODECS ===")
for command in ["mpv", "vlc", "ffmpeg", "yt-dlp", "ani-cli"]:
    path = shutil.which(command, path=os.path.expanduser("~/.local/bin")) or shutil.which(command)
    print(f"{command}: {path or 'não encontrado'}")
for command in [["mpv", "--version"], ["ffmpeg", "-version"]]:
    print("\n" + command[0] + ":")
    print(command_output(command).splitlines()[0])

print("\n=== NAVEGADORES ===")
for command in ["google-chrome", "google-chrome-stable", "chromium", "chromium-browser", "firefox", "brave-browser"]:
    path = shutil.which(command)
    if path:
        print(command, ":", path)
        print(command_output([command, "--version"]))

print("\n=== WIDEVINE / DRM ===")
widevine_roots = [
    Path.home() / ".config/google-chrome/WidevineCdm",
    Path.home() / ".config/chromium/WidevineCdm",
    Path.home() / ".mozilla/firefox",
    Path("/usr/lib/chromium/WidevineCdm"),
    Path("/opt/google/chrome/WidevineCdm"),
]
for root in widevine_roots:
    print(root, ":", "encontrado" if root.exists() else "não encontrado")

print("\n=== GPU ===")
print(command_output(["lspci"]))
```
